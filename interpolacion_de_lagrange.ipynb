# interpolacion de lagrange

## Teoria

si $x_0,x_1,...,x_n$ son $n+1$ numeros distintos y $f$ es una funcion cuyos valores estan dados en estos numeros existe un unico polinomio $P$
de grado a lo mas $n$ tal que
$$
\begin{aligned}
f(x_k)=P(x_k) \text{ con } k=1,2,..n
\end{aligned}
$$
donde el polinomio esta dado por
$$
\begin{aligned}
P(x)=\sum_{k=0}^{n}f(x_k)L_{nk}(x)
\end{aligned}
$$
donde $L_{nk}(x)$ esta dado por
$$
\begin{aligned}
L_{nk}(x)=\prod_{i=0,i\neq{k}}^{n}\frac{x-x_i}{x_k-x_i}
\end{aligned}
$$

## Cota de error

supongamos que $x_0,x_1,...,x_n$ son n+1 numeros distintos en $[a,b]$ y que $f\in{C^{n+1}}[a,b]$ entonces para cada $x$ en $[a,b]$ existe $\xi\in[a,b]$ tal que
$$
\begin{aligned}
f(x)=P(x)+\frac{f^{n+1}(\xi(x))}{(n+1)!}(x-x_1)(x-x_2)...(x-x_n)
\end{aligned}
$$
donde la cota del error esta dada por
$$
\begin{aligned}
|\epsilon|=|\frac{f^{n+1}(\xi(x))}{(n+1)!}||(x-x_1)(x-x_2)...(x-x_n)|
\end{aligned}
$$
donde se tiene que determinar la cota superior de
$$
\begin{aligned}
|\frac{f^{n+1}(\xi(x))}{(n+1)!}|
\end{aligned}
$$
y luego
$$
\begin{aligned}
|(x-x_1)(x-x_2)...(x-x_n)|
\end{aligned}
$$

```python
import numpy as np
import matplotlib.pyplot as plt
nodos=[2,2.75,4]
coefP=[]
def f(x):
    return 1/x
def sub_nodos(k):
    sub_nodos=[]
    for i in range(len(nodos)):
        if k!=i:
            sub_nodos.append(nodos[i])
    return sub_nodos
def DEN(k):
    DEN=1
    sub=sub_nodos(k)
    for i in sub:
        DEN=DEN*(nodos[k]-i)
    return DEN
def polinomio(k):
    for i in range(k):
        coefP.append(0)
    for i in range(k):
        sub=sub_nodos(i)
        d=DEN(i)
        print(d)
        for n in range(len(sub)+1):
            if n==0:
                                                                                                                                                                                                                                                                                                                                     coefP[0]=coefP[0]+(1/d)
            elif n==(len(sub)+1):
                C=1
                for q in sub:
                    C=C*q
                coefP[k]+=C/d
            #else:
    print(coefP)


def L_nk(k,x):
    sub=sub_nodos(k)
    L=1
    for i in sub:
        L=L*((x-i)/(nodos[k]-i))
    return(L)
def P(x):
    P=0
    for k in range(len(nodos)):
        P+=f(nodos[k])*L_nk(k,x)
    polinomio(len(nodos))
    print(P)
P(3)
```
