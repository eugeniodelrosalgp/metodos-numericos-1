# Iteracion de Punto Fijo

En este notebook estare explicando la teoria y un ejeplo de codigo acerca de la aproximacion de raizes utilizando el metodo de **iteracion de punto fijo**


## teoria


Primero definamos un punto fijo, este es un $p$ tal que $g(p)=p$\\
por lo que si queremos encontrar una raiz $f(p)=0$ entonces podemos definir una funcion
$$
\begin{aligned}
g(x)=x-af(x)
\\
\forall  a \in R
\end{aligned}
$$
esto nos ayuda ya que si $p$ es un punto fijo de $g$ entonces
$$
\begin{aligned}
f(x)=x-g(x)
\end{aligned}
$$

de esta manera veamos un teorema el cual una vez que proponemos una $g$ nos asegura que esta tiene un punto fijo $p$

**teorema**

si
$$
\begin{aligned}
g\in C[a,b] \text{ y }
g(x)\in[a,b]
\text{ entonces g tiene un punto fijo en } [a,b]
\end{aligned}
$$

y si ademas suponemos que existen $g\prime(x)$ en $(a,b)$ y $k<1$ tal que
$$
\begin{aligned}
\lvert g\prime(x) \rvert \leq k \text{, } \forall x \in(a,b)
\end{aligned}
$$
entonces, para cualquier $p_0\in[a,b]$ la sucesion defiida por
$$
\begin{aligned}
p_n+1=g(p_n)\text{, } \forall n> 1
\end{aligned}
$$
converge al unico punto fijo $p\in [a,b]$

## Ejemplo

Sea $f(x)=x^3+4x^2-10$ muestre que $f(x)$ tiene una raiz en el intervalo $[1,2]$

asi proponemos varias $g(x)$ las cuales puede que sirvan para este caso
<ol>
    <li>$g_1(x)=x-x^3-4x^2+10$</li>
    <li>$g_2(x)=(\frac{10}{x}-4x)^\frac{1}{2}$</li>    
    <li>$g_3(x)=\frac{1}{2}(10-x^3)^\frac{1}{2}$</li>
    <li>$g_4(x)=x-\frac{x^3+4x^2-10}{3x^2+8x}$</li>
</ol>

ya que tenemos nuestras $g(x)$ falta que cumplan la condicion de que la imagen este contenida en el dominio

de manera que

$g_1(1)=6$ y $g_1(2)=-12$  por lo que $g_1$ **no** converge

$g_2(1)=\sqrt{6}$ y $g_2(2)=indefinida$ por lo que $g_2$ **no** converge

$g_3(1)=\frac{3}{2}$ y $g_3(2)=0.7071067812$ por lo que $g_3$ **no** converge

$g_4(1)=\frac{16}{11}$ y $g_4(2)=\frac{3}{2}$ por lo que $g_4$ **si** converge

como $g_4$ converge tomemos esta para el ejemplo de programacion
de esta manera tomamos a $p_0=\frac{1+2}{2}=\frac{3}{2}$
asi tenemos que la sucecion sera
$$
\begin{aligned}
p_1=g_4(p_0)
\end{aligned}
$$

```python
import numpy as np
import matplotlib.pyplot as plt
def f(x):
    return x**3+4*x**2-10
a=1
b=2
xx=np.linspace(a-2,b+2,50)
plt.plot(xx,f(xx))
plt.axhline(y=0,color="yellow")
plt.axvline(x=0,color="green")
```

```python
def g(x):
    return (1/2)*np.sqrt(10-x**3)
def h(x):
    return x
yy=np.linspace(a-2,b+2,50)
zz=np.linspace(a-2,b+2,50)
plt.plot(yy,g(yy))
plt.plot(zz,h(zz))
plt.axhline(y=0,color="yellow")
plt.axvline(x=0,color="green")
```

```python
p0=(a+b)/2
error=100
tol=0.0000000001
n=0
pi_1=g(p0)
print("# iter\t\t $p_n$ \t\t $g(p_n)$ \t\t error")
print("{0} \t\t {1:3.4f} \t {2:3.4f} \t {3:3.4f}".format(n, p0, g(pi_1),  error ))
while error>tol and n<100:
    pi=g(pi_1)
    error=np.abs(pi-pi_1)
    n+=1
    print("{0} \t\t {1:3.10f} \t {2:3.10f} \t {3:3.10f}".format(n, pi_1, g(pi_1),  error ))
    pi_1=pi
print (pi_1)
```
